# Payroll Ingestion Pipeline

### What this delivers

An authenticated connection from Snowflake to the payroll folder in SharePoint, retrieval of the submitted files, and governance of the credentials used to do it so they are never held outside Snowflake.

### In Scope
- Connecting and authenticating to the payroll folder
- Listing what is there and retrieving files
- Detecting that a file has changed since it was last seen
- Storing the credentials and controlling who can read them
